# Statistics · Day 1 Exercise
**Task:** Use Python to summarize a small business dataset and identify what statistics can answer.

**Dataset:** Synthetic Sales

````python
import pandas as pd

# Synthetic Sales dataset (small business: 2 regions, 14 days)
data = {
    "day": list(range(1, 15)),
    "region": ["Chennai","Chennai","Chennai","Chennai","Chennai","Chennai","Chennai",
               "Coimbatore","Coimbatore","Coimbatore","Coimbatore","Coimbatore","Coimbatore","Coimbatore"],
    "sales_units": [42, 55, 38, 61, 47, 90, 35, 28, 33, 31, 40, 29, 65, 30],
    "revenue": [8250, 11300, 7450, 12500, 9150, 17800, 7100,
                5750, 6450, 6300, 8200, 5650, 12800, 6150],
}
df = pd.DataFrame(data)
df
````

````python
# Descriptive statistics: what's typical, and how spread out is it?
df["revenue"].describe().round(1)
````

````python
# Region-wise comparison: which region performs better, and how consistently?
region_summary = df.groupby("region")["sales_units"].agg(["mean", "median", "std", "min", "max"])
region_summary.round(1)
````

````python
# Relationship check: do units sold and revenue move together?
corr = df["sales_units"].corr(df["revenue"])
print(f"Correlation: {corr:.3f}")
````

````python
# Best-performing day
top_day = df.loc[df["sales_units"].idxmax()]
top_day
````

## Interpretation

Descriptive statistics show Chennai sells more on average (52.6 units/day) than Coimbatore (36.6) but is also more volatile — its standard deviation (18.9) is nearly 50% higher, meaning Chennai's sales swing further from day to day. The near-perfect correlation (0.998) between units sold and revenue confirms that revenue is driven almost entirely by volume, not by changing prices, which is itself a useful business finding. This is the core of what statistics answers with small datasets: what's typical, how much it varies, which days or regions stand out, and whether two things actually move together — all before any AI model is needed.
