## Why Statistics Matters

**Exercise:** Compare two business groups using summary statistics.

**Dataset:** Synthetic Sales — 30 days of daily sales (in ₹ thousands) for two stores.

**Goal:** Don't just compare means — compare spread (standard deviation) too. Two groups can average almost the same and still behave very differently.


```python
import numpy as np

np.random.seed(42)

# Synthetic Sales dataset
group_a = np.round(np.random.normal(loc=55, scale=8, size=30), 1)   # Group A: steady mid-size store
group_b = np.round(np.random.normal(loc=60, scale=18, size=30), 1)  # Group B: volatile, higher-traffic store

print("Group A sales:", group_a.tolist())
print("Group B sales:", group_b.tolist())
```

```python
for name, g in [("Group A", group_a), ("Group B", group_b)]:
    print(f"\n{name}")
    print("Mean:    ", round(np.mean(g), 1))
    print("Median:  ", round(np.median(g), 1))
    print("Std Dev: ", round(np.std(g, ddof=1), 1))
    print("Min/Max: ", np.min(g), "/", np.max(g))
```

### 📊 Result

| Metric   | Group A | Group B |
|----------|---------|---------|
| Mean     | 53.5    | 57.8    |
| Median   | 53.1    | 58.8    |
| Std Dev  | 7.2     | 16.8    |
| Min      | 39.7    | 24.7    |
| Max      | 67.6    | 93.3    |

### 💡 Interpretation

Group B earns more on average, but its standard deviation is over twice Group A's — meaning its daily sales swing wildly (from ₹24.7k to ₹93.3k) while Group A stays steady (₹39.7k–₹67.6k). If you only looked at the mean, you'd think Group B is simply "better." Looking at spread tells the real story: Group A is the reliable performer, Group B is the volatile one.


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 4))
plt.boxplot([group_a, group_b], labels=["Group A", "Group B"])
plt.title("Daily Sales Spread: Group A vs Group B")
plt.ylabel("Sales (₹ thousands)")
plt.grid(axis="y", alpha=0.3)
plt.show()
```
